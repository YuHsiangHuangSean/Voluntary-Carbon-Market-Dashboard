```python
column_list = [
    "Project ID",
    "Project Name",
    "Voluntary Registry",
    "Scope",
    "Type",
    "Reduction / Removal",
    "Methodology / Protocol",
    "Region",
    "Country",
    "Project Developer",
    "Certifications"
]
year_list = range(1996, 2027)

# Vintage year
df_vintage = df[column_list + [year for year in year_list]]

# Issue year
# Verra may issue credits for the same projects in multiple batches. In the raw data, it is assumed that Verra issues all credits in the first batch
df_issue = df[column_list + [str(year + 0.3) for year in year_list]]

# Retirement or cancellation year
df_retirement_cancellation = df[column_list + [str(year + 0.1) for year in year_list]]
```

```python
def df_overview(df):
    print(df.describe())
    print("")
    print(df.describe(include="object"))
    print("")
    print(df.info())
    print("")
    print(df.head(10))
```

```python
df_overview(df_vintage)
# no null that needs to be dealt with
# no duplicate that needs to be dealt with (based on Project ID)
# 2023-2026 have to be converted to int64
```

```python
df_overview(df_issue)
# no null that needs to be dealt with
# no duplicate that needs to be dealt with (based on Project ID)
# 2023-2026 have to be converted to int64
```

```python
df_overview(df_retirement_cancellation)
# no null that needs to be dealt with
# no duplicate that needs to be dealt with (based on Project ID)
# 2024-2026 have to be converted to int64
```
